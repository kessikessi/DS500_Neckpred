```python
import ast
import re
import urllib.request

station_id = "00086"
url = "https://www.hvz.baden-wuerttemberg.de/js/hvz_peg_stmn.js"

with urllib.request.urlopen(url) as response:
    javascript = response.read().decode("utf-8")

# Each station is represented by one JavaScript array on its own line.
match = re.search(
    rf"^\s*\['{re.escape(station_id)}'.*$",
    javascript,
    flags=re.MULTILINE,
)

if match is None:
    raise RuntimeError(f"Station {station_id} was not found")

# The station rows use Python-compatible list syntax.
row = ast.literal_eval(match.group(0).strip().rstrip(","))

pegeldata = {
    "station_id": row[0],
    "station": row[1],
    "river": row[2],
    "water_level": float(row[4]),
    "water_level_unit": row[5],
    "water_level_time": row[6],
    "discharge": float(row[7]),
    "discharge_unit": row[8],
    "discharge_time": row[9],
}

pegeldata
```
